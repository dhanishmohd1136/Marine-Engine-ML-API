# Marine Engine Data Preprocessing & Feature Engineering
Demonstrates feature scaling, domain thermodynamic engineering, and sensor range validation.


```python
import pandas as pd
import numpy as np
from backend.src.preprocessing import EnginePreprocessor, FEATURE_COLUMNS, SENSOR_LIMITS

df = pd.read_csv('../data/raw/marine_engine_data.csv')
print(f'Loaded {len(df)} raw telemetry records')
```

## 1. Domain Feature Engineering
We compute key thermodynamic ratios:
- `tc_pressure_ratio`: Scavenge pressure per 10k TC RPM
- `exhaust_to_load_ratio`: Thermal balance index
- `cooling_temp_delta`: Cooling water jacket outlet vs ambient
- `fuel_temp_deviation`: Viscosity control deviation


```python
preprocessor = EnginePreprocessor()
X_raw = df[FEATURE_COLUMNS]
X_scaled = preprocessor.fit_transform(X_raw)

print('Engineered Features:', preprocessor.engineered_feature_names)
print('All Features:', preprocessor.all_feature_names)
print('Scaled matrix shape:', X_scaled.shape)
```
