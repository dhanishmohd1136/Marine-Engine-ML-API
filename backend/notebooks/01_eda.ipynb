# Marine Propulsion Engine: Exploratory Data Analysis (EDA)
This notebook explores maritime telemetry data from a 2-stroke low-speed propulsion diesel engine (MAN B&W 6S50ME).


```python
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from backend.src.preprocessing import FEATURE_COLUMNS, ENGINE_CONDITIONS
from backend.src.train import generate_synthetic_marine_data

# Load raw dataset or generate if missing
data_path = '../data/raw/marine_engine_data.csv'
if os.path.exists(data_path):
    df = pd.read_csv(data_path)
else:
    df = generate_synthetic_marine_data(n_samples=2500)

print(f'Dataset Shape: {df.shape}')
df.head()
```

## 1. Summary Statistics


```python
df.describe().T
```

## 2. Operating Condition Class Breakdown


```python
condition_counts = df['condition_label'].value_counts()
print(condition_counts)

plt.figure(figsize=(8, 4))
sns.barplot(x=condition_counts.index, y=condition_counts.values, palette='viridis')
plt.title('Engine Operating Condition Distribution')
plt.xticks(rotation=25)
plt.ylabel('Sample Count')
plt.tight_layout()
plt.show()
```

## 3. Propeller Law: Engine Load vs RPM


```python
plt.figure(figsize=(8, 5))
sns.scatterplot(data=df, x='engine_rpm', y='engine_load_pct', hue='condition_label', alpha=0.7)
plt.title('Propeller Law Correlation (RPM vs Load %)')
plt.xlabel('Crankshaft RPM')
plt.ylabel('Engine Load (% MCR)')
plt.grid(True, linestyle='--', alpha=0.5)
plt.show()
```

## 4. Fuel Oil Consumption & SFOC Curves


```python
fig, axes = plt.subplots(1, 2, figsize=(14, 5))
sns.scatterplot(data=df, x='engine_load_pct', y='fuel_consumption_kg_h', hue='condition_label', ax=axes[0])
axes[0].set_title('Fuel Mass Flow (kg/h) vs Engine Load (%)')
axes[0].grid(True, alpha=0.3)

sns.scatterplot(data=df, x='engine_load_pct', y='sfoc_g_kwh', hue='condition_label', ax=axes[1])
axes[1].set_title('SFOC (g/kWh) Characteristic Bathtub Curve')
axes[1].grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```
